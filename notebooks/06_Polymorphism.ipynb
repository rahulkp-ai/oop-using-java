## Theory

```java
// Runtime polymorphism demo
abstract class Shape {
    abstract double area();
    abstract String name();
    
    void describe() {
        System.out.printf("%s: area = %.4f%n", name(), area());
    }
}

class Circle extends Shape {
    double r;
    Circle(double r) { this.r = r; }
    @Override double area() { return Math.PI * r * r; }
    @Override String name() { return "Circle(r=" + r + ")"; }
}

class Rectangle extends Shape {
    double w, h;
    Rectangle(double w, double h) { this.w = w; this.h = h; }
    @Override double area() { return w * h; }
    @Override String name() { return "Rectangle(" + w + "x" + h + ")"; }
}

class Triangle extends Shape {
    double b, h;
    Triangle(double b, double h) { this.b = b; this.h = h; }
    @Override double area() { return 0.5 * b * h; }
    @Override String name() { return "Triangle(b=" + b + ",h=" + h + ")"; }
}

// Polymorphic processing
Shape[] shapes = { new Circle(5), new Rectangle(4, 6), new Triangle(3, 8) };
for (Shape s : shapes) {
    s.describe(); // correct method called based on actual type
}
```

## Compile-time Polymorphism

```java
// Compile-time polymorphism (overloading)
class Printer {
    void print(String msg)        { System.out.println("String: " + msg); }
    void print(int num)           { System.out.println("Int: " + num); }
    void print(double d)          { System.out.println("Double: " + d); }
    void print(String s, int n)   { System.out.println("S+I: " + s + ", " + n); }
    void print(Object o)          { System.out.println("Object: " + o); }
}

Printer p = new Printer();
p.print("Hello");       // resolved at compile time
p.print(42);            // resolved at compile time
p.print(3.14);          // resolved at compile time
p.print("Java", 17);   // resolved at compile time
```

## Mini Challenge
Create a `Notification` abstract class with subclasses: `EmailNotification`, `SMSNotification`, `PushNotification`. Process a `List<Notification>` polymorphically.

```java
import java.util.ArrayList;
import java.util.List;

// 1. Abstract Base Class
abstract class Notification {
    private String recipient;
    private String message;

    public Notification(String recipient, String message) {
        this.recipient = recipient;
        this.message = message;
    }

    public String getRecipient() {
        return recipient;
    }

    public String getMessage() {
        return message;
    }

    // Abstract method to be implemented by subclasses
    public abstract void send();
}

// 2. Subclass: Email
class EmailNotification extends Notification {
    private String subject;

    public EmailNotification(String recipient, String subject, String message) {
        super(recipient, message);
        this.subject = subject;
    }

    @Override
    public void send() {
        System.out.println("   Sending EMAIL to: " + getRecipient());
        System.out.println("   Subject: " + subject);
        System.out.println("   Body: " + getMessage());
        System.out.println("------------------------------------------");
    }
}

// 3. Subclass: SMS
class SMSNotification extends Notification {
    public SMSNotification(String recipient, String message) {
        super(recipient, message);
    }

    @Override
    public void send() {
        System.out.println("   Sending SMS to: " + getRecipient());
        System.out.println("   Text: " + getMessage());
        System.out.println("------------------------------------------");
    }
}

// 4. Subclass: Push Notification
class PushNotification extends Notification {
    private String deviceToken;

    public PushNotification(String recipient, String deviceToken, String message) {
        super(recipient, message);
        this.deviceToken = deviceToken;
    }

    @Override
    public void send() {
        System.out.println("   Sending PUSH NOTIFICATION to User: " + getRecipient());
        System.out.println("   Device Token: " + deviceToken);
        System.out.println("   Payload: " + getMessage());
        System.out.println("------------------------------------------");
    }
}

// 5. Polymorphic Processing (The Notebook Execution)
List<Notification> notifications = new ArrayList<>();

notifications.add(new EmailNotification("alice@example.com", "Welcome!", "Thanks for signing up to our service."));
notifications.add(new SMSNotification("+15550199", "Your OTP code is 4821. It expires in 5 mins."));
notifications.add(new PushNotification("User_9821", "apns-tok_z89f3x", "Your order has been shipped!"));

System.out.println("=== Processing Notifications Polymorphically ===\n");

// Polmorphic loop triggers the specific send() implementation for each object type
for (Notification notification : notifications) {
    notification.send();
}
```
