# 09 — Composition and Aggregation

## Objectives
- Distinguish between composition (strong ownership) and aggregation (weak ownership)
- Apply HAS-A relationships in design
- Know when to prefer composition over inheritance

## HAS-A vs IS-A
| Relationship | Type | Example |
|-------------|------|---------|
| IS-A | Inheritance | Dog IS-A Animal |
| HAS-A (strong) | Composition | Car HAS-A Engine (Engine dies with Car) |
| HAS-A (weak) | Aggregation | Department HAS-A Professor (Professor survives) |

```java
// COMPOSITION: Engine cannot exist without Car
class Engine {
    private String type;
    private int hp;
    private boolean running;
    
    Engine(String type, int hp) { this.type = type; this.hp = hp; }
    void start() { running = true; System.out.println("[Engine:" + type + "] Started - " + hp + "HP"); }
    void stop() { running = false; System.out.println("[Engine:" + type + "] Stopped"); }
    boolean isRunning() { return running; }
}

class Car {
    private String make;
    private final Engine engine; // Car OWNS and CREATES the engine
    
    Car(String make, String engineType, int hp) {
        this.make = make;
        this.engine = new Engine(engineType, hp); // created here, dies with Car
    }
    
    void start() { engine.start(); System.out.println(make + " ready to go!"); }
    void stop()  { engine.stop(); System.out.println(make + " parked."); }
}

// AGGREGATION: Professor exists independently
class Professor {
    String name, specialization;
    Professor(String n, String s) { name = n; specialization = s; }
    void teach() { System.out.println("Prof. " + name + " teaching " + specialization); }
}

class Department {
    String name;
    java.util.List<Professor> professors = new java.util.ArrayList<>();
    
    Department(String name) { this.name = name; }
    void addProfessor(Professor p) { professors.add(p); } // Professor already exists
    void removeProfessor(Professor p) { professors.remove(p); } // Professor still exists!
}

// Demo
Car bmw = new Car("BMW M3", "TwinTurbo Inline-6", 510);
bmw.start();
bmw.stop();

System.out.println();

Professor p1 = new Professor("Dr. Ramesh", "Machine Learning");
Professor p2 = new Professor("Dr. Priya", "Data Structures");

Department cs = new Department("Computer Science");
cs.addProfessor(p1);
cs.addProfessor(p2);
System.out.println("CS Dept has " + cs.professors.size() + " professors");

cs.removeProfessor(p1);
System.out.println("After removal: " + cs.professors.size() + " professors");
p1.teach(); // p1 still exists! (aggregation)
```

## Mini Challenge
Design a `Laptop` composed of: `Processor`, `RAM`, `Battery`, `Display`. The laptop controls their lifecycle.

```java
// Define the components first
class Processor {
    private String model;

    public Processor(String model) {
        this.model = model;
    }
    
    public void boot() {
        System.out.println("Processor " + model + " is initializing...");
    }
}

class RAM {
    private int sizeGB;

    public RAM(int sizeGB) {
        this.sizeGB = sizeGB;
    }

    public void load() {
        System.out.println("Loading system files into " + sizeGB + "GB RAM...");
    }
}

class Battery {
    private int capacitymAh;

    public Battery(int capacitymAh) {
        this.capacitymAh = capacitymAh;
    }

    public void discharge() {
        System.out.println("Drawing power from " + capacitymAh + "mAh battery.");
    }
}

class Display {
    private String resolution;

    public Display(String resolution) {
        this.resolution = resolution;
    }

    public void turnOn() {
        System.out.println("Display rendering at " + resolution + ".");
    }
}

// Define the Laptop (The Composite Object)
class Laptop {
    // The components are final because they are tightly coupled to the Laptop's lifecycle
    private final Processor processor;
    private final RAM ram;
    private final Battery battery;
    private final Display display;

    // The Laptop controls the instantiation (lifecycle) of its parts
    public Laptop(String cpuModel, int ramSize, int batteryCap, String screenRes) {
        this.processor = new Processor(cpuModel);
        this.ram = new RAM(ramSize);
        this.battery = new Battery(batteryCap);
        this.display = new Display(screenRes);
    }

    public void startLaptop() {
        System.out.println("--- Powering on the Laptop ---");
        battery.discharge();
        processor.boot();
        ram.load();
        display.turnOn();
        System.out.println("Laptop is ready to use!\n");
    }
}

// Execution Cell in Jupyter Notebook
Laptop myLaptop = new Laptop("Intel i7", 16, 5000, "4K UHD");
myLaptop.startLaptop();
```
