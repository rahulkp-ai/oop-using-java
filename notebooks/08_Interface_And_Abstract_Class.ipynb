# 08 — Interfaces and Abstract Classes

## Objectives
- Master Java interfaces (100% abstraction contract)
- Compare interfaces vs abstract classes
- Use default and static methods in interfaces
- Implement multiple interfaces

## Key Distinction
| Feature | Interface | Abstract Class |
|---------|-----------|----------------|
| Multiple inheritance | Yes | No |
| Constructor | No | Yes |
| Fields | `public static final` only | Any type |
| Methods | abstract + default + static | abstract + concrete |

```java
// Interface with default method
interface Printable {
    void print();                         // abstract
    default void printWithBorder() {      // default (Java 8+)
        System.out.println("=".repeat(40));
        print();
        System.out.println("=".repeat(40));
    }
    static void info() {                  // static method
        System.out.println("Printable interface v1.0");
    }
}

interface Saveable {
    boolean save(String destination);
}

// Implementing multiple interfaces
class Invoice implements Printable, Saveable {
    private String invoiceId;
    private double amount;
    
    Invoice(String id, double amount) { this.invoiceId = id; this.amount = amount; }
    
    @Override
    public void print() {
        System.out.println("Invoice: " + invoiceId + " | Amount: INR " + amount);
    }
    
    @Override
    public boolean save(String dest) {
        System.out.println("Saving invoice " + invoiceId + " to: " + dest);
        return true;
    }
}

Invoice inv = new Invoice("INV-2024-001", 15000.0);
inv.print();
inv.printWithBorder();  // default method
inv.save("/invoices/");
Printable.info();       // static method
```

## Mini Challenge
Design a `Vehicle` interface hierarchy: `Vehicle` → `Motorized`, `NonMotorized`. Create `Car`, `Bicycle` implementations.

```java
// Core Interfaces

interface Vehicle {
    void start();
    void stop();
    default void honk() {
        System.out.println("Beep beep!");
    }
}

interface Motorized extends Vehicle {
    void refuel(double liters);
    void checkEngine();
}

interface NonMotorized extends Vehicle {
    void inspectChain();
}
```

```java
// Car Implementation

class Car implements Motorized {
    private String model;
    private double fuelLevel = 50.0; // percentage

    public Car(String model) {
        this.model = model;
    }

    @Override
    public void start() {
        System.out.println(model + " engine roars to life. Vroom!");
    }

    @Override
    public void stop() {
        System.out.println(model + " engine shuts off.");
    }

    @Override
    public void refuel(double liters) {
        System.out.println("Adding " + liters + "L of fuel to the " + model + ".");
        this.fuelLevel = Math.min(100.0, this.fuelLevel + (liters * 2)); 
    }

    @Override
    public void checkEngine() {
        System.out.println(model + " diagnostics: Fuel is at " + fuelLevel + "%. All systems nominal.");
    }
}
```

```java
// Bicycle Implementation

class Bicycle implements NonMotorized {
    private String type;
    private boolean chainLubricated = true;

    public Bicycle(String type) {
        this.type = type;
    }

    @Override
    public void start() {
        System.out.println("Hop on the " + type + " bike and start pedaling!");
    }

    @Override
    public void stop() {
        System.out.println("Squeezing the handbrakes. The " + type + " bike comes to a halt.");
    }

    @Override
    public void inspectChain() {
        System.out.println("Checking chain: " + (chainLubricated ? "Clean and greased." : "Dry! Needs oil."));
    }
    
    // Overriding default method for a more appropriate bicycle sound
    @Override
    public void honk() {
        System.out.println("Ring ring!");
    }
}
```

```java
// Testing the Hierarchy

System.out.println("--- Testing Car (Motorized) ---");
Motorized myCar = new Car("Tesla Model S (Hybrid Edition)");
myCar.start();
myCar.refuel(20);
myCar.checkEngine();
myCar.honk();
myCar.stop();

System.out.println("\n--- Testing Bicycle (NonMotorized) ---");
NonMotorized myBike = new Bicycle("Mountain");
myBike.start();
myBike.inspectChain();
myBike.honk();
myBike.stop();
```
