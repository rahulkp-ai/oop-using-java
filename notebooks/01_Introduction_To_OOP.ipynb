## What is OOP?
OOP is a programming paradigm that organizes software design around **objects** — entities that bundle **state** (data/fields) and **behavior** (methods/functions).

### Real World → Object Mapping
| Real World | OOP Concept |
|-----------|-------------|
| Blueprint of a house | Class |
| Actual house built | Object |
| Color, size, floors | Fields |
| Open door, lock | Methods |

## The 4 Pillars

```java
// Procedural approach (what NOT to do)
String name1 = "Ali";
int age1 = 22;
String major1 = "CS";

String name2 = "Dony";
int age2 = 25;
String major2 = "IT";

// Hard to manage, no encapsulation, no resue!

System.out.println("Procedural: "+ name1 + "," + age1 + "," + major1);
System.out.println("Procedural: "+ name2 + "," + age2 + "," + major2);
```

```java
// OOP approach (bundled, reusable, maintainable)
class Student {
    private String name;
    private int age;
    private String major;
    
    public Student(String name, int age, String major) {
        this.name = name;
        this.age = age;
        this.major = major;
    }
    
    public void introduce() {
        System.out.println("Hi! I am " + name + ", age " + age + ", studying " + major);
    }
    
    public boolean isAdult() { return age >= 18; }
}

Student s1 = new Student("Alice", 22, "CS");
Student s2 = new Student("Bob", 25, "IT");

s1.introduce();
s2.introduce();
System.out.println("s1 is adult: " + s1.isAdult());
```

## 4 Pillars Summary

| Pillar | Keyword | Description |
|--------|---------|-------------|
| Encapsulation | `private`, getters/setters | Protect data |
| Inheritance | `extends` | Reuse parent class |
| Polymorphism | `@Override` | Many forms |
| Abstraction | `abstract`, `interface` | Hide details |

## Mini Challenge
Model a `BankAccount` class with `accountNumber`, `balance`, and methods `deposit()` and `withdraw()`. Ensure balance never goes negative.

```java
public class BankAccount {
    // Private fields to ensure encapsulation
    private String accountNumber;
    private double balance;

    // Constructor to initialize the account
    public BankAccount(String accountNumber, double initialBalance) {
        this.accountNumber = accountNumber;
        // Ensure the account isn't opened with a negative balance
        if (initialBalance >= 0) {
            this.balance = initialBalance;
        } else {
            this.balance = 0;
            System.out.println("Initial balance cannot be negative. Setting balance to $0.0");
        }
    }

    // Method to deposit money (must be a positive amount)
    public void deposit(double amount) {
        if (amount > 0) {
            balance += amount;
            System.out.println("Successfully deposited: $" + amount);
        } else {
            System.out.println("Invalid deposit amount. Amount must be positive.");
        }
    }

    // Method to withdraw money (checks if funds are sufficient)
    public void withdraw(double amount) {
        if (amount <= 0) {
            System.out.println("Invalid withdrawal amount. Amount must be positive.");
        } else if (amount > balance) {
            System.out.println("Transaction denied: Insufficient funds. Current balance: $" + balance);
        } else {
            balance -= amount;
            System.out.println("Successfully withdrew: $" + amount);
        }
    }

    // Getter method to safely view the balance
    public double getBalance() {
        return balance;
    }

    // Getter method for the account number
    public String getAccountNumber() {
        return accountNumber;
    }
}
```
