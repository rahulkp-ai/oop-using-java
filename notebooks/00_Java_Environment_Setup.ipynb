## Quick Verification

````java
// Verify Java version in terminal:
// java -version
// javac -version
// mvn -version

// HelloWorld in Java
public class HelloWorld {
    public static void main(String[] args) {
        System.out.println("Java Environment Ready!");
        System.out.println("Java version: " + System.getProperty("java.version"));
        System.out.println("JVM name: " + System.getProperty("java.vm.name"));
    }
}
HelloWorld.main(null);
````

## Mini Challenge
Create a `HelloStudent` class that prints your name and the current date using `java.time.LocalDate.now()`.

````java
import java.time.LocalDate;
public class HelloStudent{
    public static void main(String[] args){
        String name = "RAHUL KP KURUP";
        LocalDate currentDate = LocalDate.now();
        System.out.println("Hello, my name is " + name + "!");
        System.out.println("Today's date is: " + currentDate);
    }
}
HelloStudent.main(null);
````
