# 17 — Lambda Expressions & Functional Interfaces

## Objectives
- Write lambda expressions for functional interfaces
- Use built-in functional interfaces: `Predicate`, `Function`, `Consumer`, `Supplier`
- Apply method references
- Use lambdas with Streams

## Lambda Syntax
```java
(params) -> expression
(params) -> { statements; }
() -> expression
```

````java
import java.util.*;
import java.util.function.*;
import java.util.stream.*;

// Functional interfaces
Predicate<Integer> isEven = n -> n % 2 == 0;
Predicate<String> isLong  = s -> s.length() > 5;
Function<String, Integer> strLen = String::length;     // method reference
Function<Integer, String> intToStr = n -> "Number: " + n;
Consumer<String> printer = System.out::println;        // method reference
Supplier<List<String>> listFactory = ArrayList::new;   // constructor reference

System.out.println("Is 4 even? " + isEven.test(4));
System.out.println("Is 'Java' long? " + isLong.test("Java"));
System.out.println("Length of 'Generics': " + strLen.apply("Generics"));

// Compose functions
Function<String, String> upperAndTrim = String::trim;
Function<String, Integer> pipeline = upperAndTrim.andThen(strLen);
System.out.println("Pipeline ' hello world ': " + pipeline.apply(" hello world "));

// Lambda with Streams
List<String> names = Arrays.asList("Alice", "Bob", "Charlie", "Diana", "Eve", "Frank");

System.out.println("\nNames with length > 4, sorted:");
names.stream()
    .filter(isLong)
    .sorted()
    .map(String::toUpperCase)
    .forEach(printer);

// Grouping
Map<Integer, List<String>> byLength = names.stream()
    .collect(Collectors.groupingBy(String::length));
System.out.println("\nGrouped by length: " + byLength);

// Reduce
int totalLength = names.stream().mapToInt(String::length).reduce(0, Integer::sum);
System.out.println("Total chars: " + totalLength);
````


## Mini Challenge
Using lambdas and streams: read a list of `Product` objects, filter those with price > 500, group by category, and find the average price per category.

````java
import java.util.*;
import java.util.stream.*;

// 1. Define the Product class blueprint
class Product {
    private String name;
    private String category;
    private double price;

    public Product(String name, String category, double price) {
        this.name = name;
        this.category = category;
        this.price = price;
    }

    public String getCategory() { return category; }
    public double getPrice() { return price; }

    @Override
    public String toString() {
        return name + " (" + price + ")";
    }
}

// 2. Setup mock data
List<Product> products = Arrays.asList(
    new Product("MacBook Pro", "Electronics", 1999.99),
    new Product("iPhone 15", "Electronics", 799.99),
    new Product("Keychron Keyboard", "Electronics", 90.00),
    new Product("Desk Chair", "Furniture", 250.00),
    new Product("Standing Desk", "Furniture", 650.00),
    new Product("Java Hoodie", "Apparel", 45.00),
    new Product("Leather Jacket", "Apparel", 550.00)
);

// 3. The Stream Pipeline Challenge
Map<String, Double> avgPriceByCategory = products.stream()
    .filter(p -> p.getPrice() > 500) // Filter prices > 500
    .collect(Collectors.groupingBy(
        Product::getCategory,        // Group by category
        Collectors.averagingDouble(Product::getPrice) // Downstream collector for average
    ));

// Print the result
System.out.println("Average price per category (for products > $500):");
avgPriceByCategory.forEach((category, avgPrice) -> 
    System.out.printf("%s: $%.2f%n", category, avgPrice)
);
````
