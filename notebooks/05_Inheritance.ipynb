## Theory + Hierarchy Diagram

```java
// Multi-level inheritance
class Animal {
    protected String name;
    protected int age;
    
    public Animal(String name, int age) {
        this.name = name;
        this.age = age;
    }
    
    public void eat() { System.out.println(name + " is eating."); }
    public void sleep() { System.out.println(name + " is sleeping."); }
    public String toString() { return name + " (" + age + " years)"; }
}

class Dog extends Animal {
    private String breed;
    
    public Dog(String name, int age, String breed) {
        super(name, age); // call parent constructor
        this.breed = breed;
    }
    
    public void bark() { System.out.println(name + " (" + breed + "): Woof!"); }
    
    @Override
    public String toString() { return super.toString() + " [" + breed + "]"; }
}

class GoldenRetriever extends Dog {
    public GoldenRetriever(String name, int age) {
        super(name, age, "Golden Retriever");
    }
    
    public void fetch() { System.out.println(name + " fetches the ball!"); }
}

GoldenRetriever buddy = new GoldenRetriever("Buddy", 3);
buddy.eat();    // from Animal
buddy.bark();   // from Dog
buddy.fetch();  // own method
System.out.println(buddy);  // overridden toString
```

## Method Overriding

```java
// Runtime polymorphism via inheritance
Animal[] animals = {
    new Dog("Rex", 4, "German Shepherd"),
    new GoldenRetriever("Max", 2)
};

System.out.println("\nAll animals eat:");
for (Animal a : animals) {
    a.eat();  // same call, same result (eat() not overridden here)
    System.out.println("  -> " + a);  // different toString per class
}
```

## Mini Challenge
Create: `Shape` → `Polygon` → `Rectangle` → `Square`
Each level adds specific behavior. Override `area()` and `perimeter()` correctly.

```java
// Base Shape Class
public abstract class Shape {
    // Abstract methods that all shapes must implement
    public abstract double area();
    public abstract double perimeter();
    
    @Override
    public String toString() {
        return String.format("%s [Area: %.2f, Perimeter: %.2f]", 
            this.getClass().getSimpleName(), area(), perimeter());
    }
}
```

```java
// Polygon Level (Adds sides tracking)
public abstract class Polygon extends Shape {
    private final int numberOfSides;

    public Polygon(int numberOfSides) {
        if (numberOfSides < 3) {
            throw new IllegalArgumentException("A polygon must have at least 3 sides.");
        }
        this.numberOfSides = numberOfSides;
    }

    public int getNumberOfSides() {
        return numberOfSides;
    }
}
```

```java
// Rectangle Level (Adds width and height)
public class Rectangle extends Polygon {
    private double width;
    private double height;

    public Rectangle(double width, double height) {
        super(4); // A rectangle always has 4 sides
        this.width = width;
        this.height = height;
    }

    @Override
    public double area() {
        return width * height;
    }

    @Override
    public double perimeter() {
        return 2 * (width + height);
    }

    // Getters and setters for specific behavior modifications
    public double getWidth() { return width; }
    public void setWidth(double width) { this.width = width; }

    public double getHeight() { return height; }
    public void setHeight(double height) { this.height = height; }
}
```

```java
// Square Level (Enforces equal sides)
public class Square extends Rectangle {
    
    public Square(double side) {
        // A square is a rectangle where width == height
        super(side, side); 
    }

    // Override setters to maintain the "Square" invariant (Liskov Substitution Principle handling)
    @Override
    public void setWidth(double width) {
        super.setWidth(width);
        super.setHeight(width);
    }

    @Override
    public void setHeight(double height) {
        super.setWidth(height);
        super.setHeight(height);
    }
    
    public double getSide() {
        return getWidth(); // Since width and height are always equal
    }
}
```

```java
// Test Execution
Shape rect = new Rectangle(5.0, 4.0);
Shape mySquare = new Square(4.0);

System.out.println(rect);
System.out.println(mySquare);

System.out.println("\n--- Testing Mutation ---");
Square dynamicSquare = new Square(5.0);
System.out.println("Original Square: " + dynamicSquare);
dynamicSquare.setWidth(8.0); 
System.out.println("After changing width to 8.0: " + dynamicSquare);
```
