## Theory

```java
class Employee {
    private String id;
    private String name;
    private String dept;
    private double salary;
    
    // Default constructor
    public Employee() {
        this("EMP-DEFAULT", "Unknown", "General", 30000);
        System.out.println("Default constructor called");
    }
    
    // Parameterized constructor
    public Employee(String id, String name, String dept, double salary) {
        this.id = id;
        this.name = name;
        this.dept = dept;
        this.salary = salary;
        System.out.println("Full constructor called for: " + name);
    }
    
    // Copy constructor
    public Employee(Employee other) {
        this(other.id + "-COPY", other.name, other.dept, other.salary);
    }
    
    public String toString() {
        return id + " | " + name + " | " + dept + " | INR " + salary;
    }
}

Employee e1 = new Employee();
Employee e2 = new Employee("E001", "Alice", "Engineering", 85000);
Employee e3 = new Employee(e2); // copy

System.out.println(e1);
System.out.println(e2);
System.out.println(e3);
```

## Constructor Chaining

```java
// Constructor chaining with this()
class Point {
    double x, y, z;
    Point() { this(0, 0, 0); }
    Point(double x, double y) { this(x, y, 0); }
    Point(double x, double y, double z) { this.x=x; this.y=y; this.z=z; }
    public String toString() { return "(" + x + "," + y + "," + z + ")"; }
}

System.out.println(new Point());
System.out.println(new Point(3, 4));
System.out.println(new Point(1, 2, 3));
```

## Mini Challenge
Create an `Order` class with three constructors: one with just orderId, one with orderId+customer, and one full constructor. Use constructor chaining.

```java
class Order {
    private String orderId;
    private String customerName;
    private double totalAmount;
    private String status;

    public Order(String orderId) {
        this(orderId, "Guest Customer");
    }

    public Order(String orderId, String customerName) {
        this(orderId, customerName, 0.0, "Pending");
    }

    public Order(String orderId, String customerName, double totalAmount, String status) {
        this.orderId = orderId;
        this.customerName = customerName;
        this.totalAmount = totalAmount;
        this.status = status;
        System.out.println("Master constructor executed for Order: " + orderId);
    }

    @Override
    public String toString() {
        return "ID: " + orderId + " | Customer: " + customerName + 
               " | Amount: INR " + totalAmount + " | Status: " + status;
    }
}
```

```java
System.out.println("--- Creating Order 1 (ID only) ---");
Order o1 = new Order("ORD-101");
System.out.println(o1);

System.out.println("\n--- Creating Order 2 (ID + Customer) ---");
Order o2 = new Order("ORD-102", "Bob");
System.out.println(o2);
```
