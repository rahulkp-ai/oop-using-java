## Why Generics?
```java
// Without generics (old way)
List list = new ArrayList();
list.add("hello");
String s = (String) list.get(0); // cast needed, ClassCastException risk!

// With generics
List<String> list = new ArrayList<>();
list.add("hello");
String s = list.get(0); // no cast, compile-time safety!
```

````java
// Generic Stack implementation
class Stack<T> {
    private Object[] data;
    private int top;
    private int capacity;
    
    @SuppressWarnings("unchecked")
    Stack(int capacity) {
        this.capacity = capacity;
        data = new Object[capacity];
        top = -1;
    }
    
    void push(T item) {
        if (top == capacity - 1) throw new RuntimeException("Stack overflow");
        data[++top] = item;
    }
    
    @SuppressWarnings("unchecked")
    T pop() {
        if (top == -1) throw new RuntimeException("Stack underflow");
        return (T) data[top--];
    }
    
    @SuppressWarnings("unchecked")
    T peek() { return top == -1 ? null : (T) data[top]; }
    
    boolean isEmpty() { return top == -1; }
    int size() { return top + 1; }
    public String toString() {
        StringBuilder sb = new StringBuilder("[");
        for (int i = 0; i <= top; i++) sb.append(data[i]).append(i < top ? ", " : "");
        return sb.append("]").toString();
    }
}

// Generic method
<T extends Comparable<T>> T findMax(T[] arr) {
    T max = arr[0];
    for (T item : arr) if (item.compareTo(max) > 0) max = item;
    return max;
}

// Integer stack
Stack<Integer> intStack = new Stack<>(5);
intStack.push(10); intStack.push(20); intStack.push(30);
System.out.println("Int Stack: " + intStack);
System.out.println("Pop: " + intStack.pop());
System.out.println("Peek: " + intStack.peek());

// String stack
Stack<String> strStack = new Stack<>(5);
strStack.push("Java"); strStack.push("OOP"); strStack.push("Generics");
System.out.println("\nStr Stack: " + strStack);

// Generic method
String[] words = {"Mango", "Apple", "Cherry", "Banana"};
Double[] prices = {99.9, 249.0, 149.5, 79.0};
System.out.println("\nMax word: " + findMax(words));
System.out.println("Max price: " + findMax(prices));
````

## Mini Challenge
Implement a generic `Pair<A,B>` and `Triple<A,B,C>` class. Create a method that swaps the elements of a `Pair`.

````java
// Generic Pair Class
class Pair<A, B> {
    private A first;
    private B second;

    public Pair(A first, B second) {
        this.first = first;
        this.second = second;
    }

    public A getFirst() { return first; }
    public void setFirst(A first) { this.first = first; }

    public B getSecond() { return second; }
    public void setSecond(B second) { this.second = second; }

    // Instance method to swap internally (only works cleanly if A and B are the same type)
    @SuppressWarnings("unchecked")
    public void swap() {
        // This cast is safe only if A and B are compatible types at runtime
        A temp = first;
        first = (A) second;
        second = (B) temp;
    }

    @Override
    public String toString() {
        return "(" + first + ", " + second + ")";
    }
}

// Generic Triple Class extending the concept
class Triple<A, B, C> {
    private A first;
    private B second;
    private C third;

    public Triple(A first, B second, C third) {
        this.first = first;
        this.second = second;
        this.third = third;
    }

    public A getFirst() { return first; }
    public B getSecond() { return second; }
    public C getThird() { return third; }

    @Override
    public String toString() {
        return "(" + first + ", " + second + ", " + third + ")";
    }
}

// Main class to demonstrate functionality
public class Main {
    
    // Generic method to swap a Pair and return a NEW Pair with reversed types
    public static <A, B> Pair<B, A> swapPair(Pair<A, B> pair) {
        return new Pair<>(pair.getSecond(), pair.getFirst());
    }

    public static void main(String[] args) {
        // 1. Testing Pair and Triple creation
        Pair<String, Integer> student = new Pair<>("Alice", 101);
        Triple<String, Integer, Double> product = new Triple<>("Laptop", 1, 899.99);

        System.out.println("Original Pair: " + student);
        System.out.println("Original Triple: " + product);
        System.out.println("---");

        // 2. Testing the generic swap method (Returns a new Pair with inverted types)
        Pair<Integer, String> swappedStudent = swapPair(student);
        System.out.println("Swapped Pair (New Instance): " + swappedStudent);

        // 3. Testing internal swap (Works smoothly when types match)
        Pair<String, String> coordinates = new Pair<>("North", "East");
        System.out.println("\nOriginal Matching Pair: " + coordinates);
        coordinates.swap();
        System.out.println("Internally Swapped Pair: " + coordinates);
    }
}
````

````java
// Cell 2: Execution
Pair<String, Integer> student = new Pair<>("Alice", 101);
Triple<String, Integer, Double> product = new Triple<>("Laptop", 1, 899.99);

System.out.println("Original Pair: " + student);
System.out.println("Original Triple: " + product);
System.out.println("---");

// Testing the generic swap method (In a notebook, the method needs to be defined 
// globally or inside a class. If defined in Main, call Main.swapPair)
Pair<Integer, String> swappedStudent = Main.swapPair(student);
System.out.println("Swapped Pair (New Instance): " + swappedStudent);

Pair<String, String> coordinates = new Pair<>("North", "East");
System.out.println("\nOriginal Matching Pair: " + coordinates);
coordinates.swap();
System.out.println("Internally Swapped Pair: " + coordinates);
````
