## Why Generics?
```java
// Without generics (old way)
List list = new ArrayList();
list.add("hello");
String s = (String) list.get(0); // cast needed, ClassCastException risk!

// With generics
List<String> list = new ArrayList<>();
list.add("hello");
String s = list.get(0); // no cast, compile-time safety!
```

````java
// Generic Stack implementation
class Stack<T> {
    private Object[] data;
    private int top;
    private int capacity;
    
    @SuppressWarnings("unchecked")
    Stack(int capacity) {
        this.capacity = capacity;
        data = new Object[capacity];
        top = -1;
    }
    
    void push(T item) {
        if (top == capacity - 1) throw new RuntimeException("Stack overflow");
        data[++top] = item;
    }
    
    @SuppressWarnings("unchecked")
    T pop() {
        if (top == -1) throw new RuntimeException("Stack underflow");
        return (T) data[top--];
    }
    
    @SuppressWarnings("unchecked")
    T peek() { return top == -1 ? null : (T) data[top]; }
    
    boolean isEmpty() { return top == -1; }
    int size() { return top + 1; }
    public String toString() {
        StringBuilder sb = new StringBuilder("[");
        for (int i = 0; i <= top; i++) sb.append(data[i]).append(i < top ? ", " : "");
        return sb.append("]").toString();
    }
}

// Generic method
<T extends Comparable<T>> T findMax(T[] arr) {
    T max = arr[0];
    for (T item : arr) if (item.compareTo(max) > 0) max = item;
    return max;
}

// Integer stack
Stack<Integer> intStack = new Stack<>(5);
intStack.push(10); intStack.push(20); intStack.push(30);
System.out.println("Int Stack: " + intStack);
System.out.println("Pop: " + intStack.pop());
System.out.println("Peek: " + intStack.peek());

// String stack
Stack<String> strStack = new Stack<>(5);
strStack.push("Java"); strStack.push("OOP"); strStack.push("Generics");
System.out.println("\nStr Stack: " + strStack);

// Generic method
String[] words = {"Mango", "Apple", "Cherry", "Banana"};
Double[] prices = {99.9, 249.0, 149.5, 79.0};
System.out.println("\nMax word: " + findMax(words));
System.out.println("Max price: " + findMax(prices));
````
