## System Design Process
```
1. Clarify requirements (functional + non-functional)
2. Identify entities (nouns → classes)
3. Identify behaviors (verbs → methods)
4. Define relationships (HAS-A, IS-A, USES)
5. Apply design patterns
6. Draw UML class diagram
7. Write code with tests
```

````java
// LLD Case Study: Parking Lot System
enum VehicleType { BIKE, CAR, BUS }
enum SpotStatus { AVAILABLE, OCCUPIED }

class Vehicle {
    private String licensePlate;
    private VehicleType type;
    Vehicle(String plate, VehicleType type) { this.licensePlate = plate; this.type = type; }
    VehicleType getType() { return type; }
    String getLicensePlate() { return licensePlate; }
}

class ParkingSpot {
    private int spotId;
    private VehicleType allowedType;
    private SpotStatus status;
    private Vehicle currentVehicle;
    
    ParkingSpot(int id, VehicleType type) {
        spotId = id; allowedType = type; status = SpotStatus.AVAILABLE;
    }
    
    boolean park(Vehicle v) {
        if (status == SpotStatus.OCCUPIED || v.getType() != allowedType) return false;
        currentVehicle = v;
        status = SpotStatus.OCCUPIED;
        System.out.printf("  Spot %d: Parked %s [%s]%n", spotId, v.getLicensePlate(), allowedType);
        return true;
    }
    
    Vehicle unpark() {
        if (status == SpotStatus.AVAILABLE) return null;
        Vehicle v = currentVehicle;
        currentVehicle = null;
        status = SpotStatus.AVAILABLE;
        System.out.printf("  Spot %d: Released %s%n", spotId, v.getLicensePlate());
        return v;
    }
    
    boolean isAvailable() { return status == SpotStatus.AVAILABLE; }
    int getSpotId() { return spotId; }
}

class ParkingLot {
    private String name;
    private java.util.List<ParkingSpot> spots = new java.util.ArrayList<>();
    
    ParkingLot(String name) { this.name = name; }
    
    void addSpot(ParkingSpot spot) { spots.add(spot); }
    
    ParkingSpot findSpot(VehicleType type) {
        return spots.stream()
            .filter(s -> s.isAvailable() && true)
            .filter(s -> {
                try { return s.park(new Vehicle("TEST", type)) ? true : false; }
                catch (Exception e) { return false; }
            })
            .findFirst().orElse(null);
    }
    
    boolean parkVehicle(Vehicle v) {
        for (ParkingSpot spot : spots) {
            if (spot.park(v)) return true;
        }
        System.out.println("  No available spot for " + v.getType());
        return false;
    }
    
    void status() {
        long available = spots.stream().filter(ParkingSpot::isAvailable).count();
        System.out.printf("  %s: %d/%d spots available%n", name, available, spots.size());
    }
}

// Demo
ParkingLot lot = new ParkingLot("Tech Park Parking");
lot.addSpot(new ParkingSpot(1, VehicleType.BIKE));
lot.addSpot(new ParkingSpot(2, VehicleType.BIKE));
lot.addSpot(new ParkingSpot(3, VehicleType.CAR));
lot.addSpot(new ParkingSpot(4, VehicleType.CAR));
lot.addSpot(new ParkingSpot(5, VehicleType.BUS));

System.out.println("=== Parking Lot System ===");
lot.status();

Vehicle bike1 = new Vehicle("KL-01-AB-1234", VehicleType.BIKE);
Vehicle car1  = new Vehicle("KA-03-MN-5678", VehicleType.CAR);
Vehicle car2  = new Vehicle("MH-12-CD-9012", VehicleType.CAR);
Vehicle car3  = new Vehicle("TN-07-EF-3456", VehicleType.CAR); // should fail

System.out.println("\nParking vehicles:");
lot.parkVehicle(bike1);
lot.parkVehicle(car1);
lot.parkVehicle(car2);
lot.parkVehicle(car3); // no spot available

System.out.println("\nAfter parking:");
lot.status();
````
