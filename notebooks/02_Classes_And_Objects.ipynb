# Theory

```java
// Class definition
class Car {
    // Fields (state)
    private String brand;
    private String model;
    private int year;
    private double price;
    
    // Constructor
    public Car(String brand, String model, int year, double price) {
        this.brand = brand;
        this.model = model;
        this.year = year;
        this.price = price;
    }
    
    // Methods (behavior)
    public String getDetails() {
        return year + " " + brand + " " + model + " @ INR " + price;
    }
    
    public void applyDiscount(double percent) {
        price = price * (1 - percent / 100);
        System.out.println("Discounted price: INR " + String.format("%.2f", price));
    }
}

// Creating objects
Car car1 = new Car("Toyota", "Camry", 2023, 2500000);
Car car2 = new Car("Honda", "City", 2023, 1200000);

System.out.println(car1.getDetails());
System.out.println(car2.getDetails());
car1.applyDiscount(10);
```

## Object Lifecycle

```java
// Object reference demo
Car a = new Car("Ford", "Mustang", 2022, 5000000);
Car b = a;  // b points to SAME object as a

System.out.println("Same object? " + (a == b));

// Creating a truly separate car
Car c = new Car("Ford", "Mustang", 2022, 5000000);
System.out.println("Different object: " + (a == c));
```

## Mini Challenge
Create a `Product` class with `id`, `name`, `price`, `stock`. Add methods to apply a discount and check availability.

```java
public class Product {
    // Instance variables (Encapsulation)
    private String id;
    private String name;
    private double price;
    private int stock;

    // Constructor
    public Product(String id, String name, double price, int stock) {
        this.id = id;
        this.name = name;
        this.price = price;
        this.stock = stock;
    }

    /**
     * Applies a percentage discount to the product price.
     * percentage The discount percentage (e.g., 10.0 for 10%)
     */
    public void applyDiscount(double percentage) {
        if (percentage > 0 && percentage <= 100) {
            this.price -= this.price * (percentage / 100.0);
        } else {
            System.out.println("Invalid discount percentage. Must be between 0 and 100.");
        }
    }

    /**
     * Checks if the product is in stock.
     * true if stock is greater than 0, false otherwise.
     */
    public boolean isAvailable() {
        return this.stock > 0;
    }

    // --- Getters and Setters ---

    public String getId() {
        return id;
    }

    public void setId(String id) {
        this.id = id;
    }

    public String getName() {
        return name;
    }

    public void setName(String name) {
        this.name = name;
    }

    public double getPrice() {
        return price;
    }

    public void setPrice(double price) {
        if (price >= 0) {
            this.price = price;
        } else {
            System.out.println("Price cannot be negative.");
        }
    }

    public int getStock() {
        return stock;
    }

    public void setStock(int stock) {
        if (stock >= 0) {
            this.stock = stock;
        } else {
            System.out.println("Stock cannot be negative.");
        }
    }
}
```

```java
// Create a new product directly
Product laptop = new Product("P101", "Wireless Mouse", 50.0, 5);

// Check initial availability
System.out.println("Is available? " + laptop.isAvailable());

// Apply a 10% discount
laptop.applyDiscount(10);
System.out.println("Price after 10% discount: " + laptop.getPrice());

// Simulate selling out the stock
laptop.setStock(0);
System.out.println("Is available after selling out? " + laptop.isAvailable());
```
